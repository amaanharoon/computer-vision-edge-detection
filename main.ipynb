### Importing Images for Keras

```python
mnist_pick = tf.keras.datasets.mnist.load_data()
```

```python
# 1. Load the MNIST dataset
(x_train, y_train), (_, _) = mnist_pick

# 2. Select the top 5 images and their corresponding labels
images = x_train[:5]
labels = y_train[:5]


# 3. Create a figure with 5 subplots to display images side-by-side
fig, axes = plt.subplots(1, 5, figsize=(15, 3))

# 4. Iterate and plot each image
for i, ax in enumerate(axes):
    # Display the NumPy array as a grayscale image
    ax.imshow(images[i], cmap='gray')
    ax.set_title(f"Label: {labels[i]}", fontsize=14)
    # Hide the axis ticks for a cleaner look
    ax.axis('off')

# Adjust layout and show the plot
plt.tight_layout()
plt.show()
```
